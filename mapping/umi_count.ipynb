```python
import os
import pandas as pd

# 设置输入目录和输出目录
input_dir = "./igblastn_out"  # 根目录
output_dir = "./igblast_out/umi/IGH"     # 输出目录

# 如果输出目录不存在，创建目录
os.makedirs(output_dir, exist_ok=True)

# 递归遍历目录及子目录，查找所有 .tsv 文件
for root, dirs, files in os.walk(input_dir):
    for tsv_file in files:
        if tsv_file.endswith(".tsv"):
            # 获取文件的完整路径
            tsv_path = os.path.join(root, tsv_file)
            
            # 读取 TSV 文件
            df = pd.read_csv(tsv_path, sep='\t')
            
            # 判断是否存在 'barcode' 列
            if 'barcode' in df.columns:
                # 提取 'barcode' 列，并按 '_' 分割，提取第二段内容作为 'umi_count'
                df['umi_count'] = df['barcode'].str.split('_').str[2]
                
                # 输出结果到新的文件，保持原始文件夹结构
                relative_path = os.path.relpath(root, input_dir)
                output_subdir = os.path.join(output_dir, relative_path)
                os.makedirs(output_subdir, exist_ok=True)
                
                output_file = os.path.join(output_subdir, tsv_file)
                df.to_csv(output_file, sep='\t', index=False)
                print(f"处理完成: {output_file}")
            else:
                print(f"警告: 文件 {tsv_file} 中没有 'sequence_id' 列，跳过该文件")
```
